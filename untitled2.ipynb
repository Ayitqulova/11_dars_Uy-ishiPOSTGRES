```python
import pandas as pd

df = pd.read_csv('/content/sample_data/4.CSV')

describe = df.describe()
mean = df.mean(numeric_only=True)
median = df.median(numeric_only=True)
max_val = df.max(numeric_only=True)
min_val = df.min(numeric_only=True)
sum_val = df.sum(numeric_only=True)
count = df.count()
std = df.std(numeric_only=True)

print(describe)
print(mean)
print(median)
print(max_val)
print(min_val)
print(sum_val)
print(count)
print(std)

describe.to_csv('describe.csv')
mean.to_csv('mean.csv')
median.to_csv('median.csv')
max_val.to_csv('max.csv')
min_val.to_csv('min.csv')
sum_val.to_csv('sum.csv')
count.to_csv('count.csv')
std.to_csv('std.csv')
```
